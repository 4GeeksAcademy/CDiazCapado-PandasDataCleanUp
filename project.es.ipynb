#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python
# direccion y precio casa mas cara
idx = ds['price'].idxmax()
casa_mas_cara_direccion = ds.loc[idx, 'address']
casa_mas_cara_precio= ds.loc[idx, 'price']
casa_mas_cara_texto = f'El precio de la casa más cara es: {casa_mas_cara_precio} USD y su direccion es: {casa_mas_cara_direccion}'
print(casa_mas_cara_texto)
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)

Este ejercicio es similar al ejercicio anterior, solo que ahora buscamos la casa con el precio más bajo. Recuerda imprimir la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle Alcalá, Nº58 es la más barata y su precio es de 12000 USD`

#### Ejercicio 04. ¿Cuantas poblaciones contiene el dataset? (★☆☆)

Cuenta el número de poblaciones únicas en la columna 'level5' e imprime el nombre de las poblaciones separadas por coma. Por ejemplo:

`> print(populations)`

`population1, population2, population3, ...`

```python

numero_poblaciones_unicas = ds['level5'].nunique()
nombres_poblaciones_unicas = ', '.join(ds['level5'].unique())
print('Poblaciones: '+ nombres_poblaciones_unicas)
numero_veces_que_aparece_arroyomolinos = ds['level5'].value_counts().get('Arroyomolinos', 0)
print(f'Número de veces que aparece Arroyomolinos: {numero_veces_que_aparece_arroyomolinos}')
#donde aparece la palabra arroyomolinos en el dataset en cualquier columna  
filas_con_arroyomolinos = ds.isin(['Arroyomolinos']).any(axis=1)
print(f'Filas donde aparece Arroyomolinos: {ds[filas_con_arroyomolinos]}')
```

#### Ejercicio 05. ¿El dataset contiene valores nulos (NAs)? (★☆☆)

Imprima un booleano (`True` o `False`) para verificar si hay valores nulos seguido de las columnas que contiene el NAs.

```python
# imprime un booleano que verifique columna por columna si hay valores nulos, seguidos del nombre de las columnas que tienen valores nulos y metelo todo en una variable que se llame columnas_con_valores_nulos

columnas_con_valores_nulos = ds.columns[ds.isnull().any()].tolist()
print(ds.isnull().values.any())
print('Columnas con valores nulos: ' + ', '.join(columnas_con_valores_nulos))  
numero_valores_nulos = ds[columnas_con_valores_nulos].isnull().sum().sum()
print(f'Número total de valores nulos en las columnas con valores nulos: {numero_valores_nulos}')
numero_total_filas_con_valores_nulos = ds[columnas_con_valores_nulos].isnull().any(axis=1).sum()
print(f'Número total de filas con valores nulos en las columnas con valores nulos: {numero_total_filas_con_valores_nulos}')
```

#### Ejercicio 06. Elimina los valores nulos (NAs) del dataset, si aplica (★★☆)

Despues de eliminar los valores nulos, compara el tamaño del DataFrame antes y después de la eliminación.


```python
# Elimina los valores nulos del dataset usa la variable columnas 
ds_limpio = ds.dropna(axis=1, how='any') # elimina columnas con valores nulos

#compara el tamano del dataset original con el dataset limpio 
print(f'Tamaño del dataset original:  {pd.read_csv("assets/real_estate.csv", sep=";").shape}')
print(f'Tamaño del dataset limpio:  {ds_limpio.shape}')
```

#### Ejercicio 07. ¿Cuál es la media de precios en la población de "Arroyomolinos (Madrid)"? (★★☆)

Imprima el valor obtenido de la columna level5.

```python
# imprime los valores de precios cuando la columna level 5 tiene de valor Arroyomolinos
print(ds_limpio.loc[ds_limpio['level5'] == 'Arroyomolinos', 'price'])
media_precios_arroyomolinos = ds_limpio.loc[ds_limpio['level5'] == 'Arroyomolinos', 'price'].mean()
print(f'La media de los precios en Arroyomolinos es: {media_precios_arroyomolinos}')

#resulta que arroyomolinos no aparece en todo el data set incluso sin limpiar ,lo comprobamos con el dataset original
filas_con_arroyomolinos = ds.isin(['Arroyomolinos']).any(axis=1)
print(f'Filas donde aparece Arroyomolinos: {ds[filas_con_arroyomolinos]}')
#asi que lo que voy a hacer es buscar la poblacion que se repita mas en level5 y comparar la evolucion de precios
from collections import Counter
poblacion_mas_frecuente = Counter(ds_limpio['level5']).most_common(1)[0][0]
print(f'La población que más se repite es: {poblacion_mas_frecuente}')
precios_poblacion_mas_frecuente = ds_limpio.loc[ds_limpio['level5'] == poblacion_mas_frecuente, 'price']
print(f'Precios en {poblacion_mas_frecuente}: {precios_poblacion_mas_frecuente.tolist()}')
media_precios_poblacion_mas_frecuente = precios_poblacion_mas_frecuente.mean()
print(f'La media de los precios en {poblacion_mas_frecuente} es: {media_precios_poblacion_mas_frecuente}')  
precio_min_poblacion_mas_frecuente_no_cero = precios_poblacion_mas_frecuente[precios_poblacion_mas_frecuente > 0].min()
print(f'El precio mínimo en {poblacion_mas_frecuente} (sin contar ceros) es: {precio_min_poblacion_mas_frecuente_no_cero}')
precio_max_poblacion_mas_frecuente = precios_poblacion_mas_frecuente.max()
print(f'El precio máximo en {poblacion_mas_frecuente} es: {precio_max_poblacion_mas_frecuente}')  
```

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)

Imprime el histograma de los precios y escribe en la celda del Markdown un breve análisis del trazado.


```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
plt.figure(figsize=(8, 5))
ax = plt.gca()
ax.xaxis.set_major_locator(ticker.MultipleLocator(300_000))
plt.xticks(rotation=45)

plt.hist(
    precios_poblacion_mas_frecuente,
    bins=50,
    edgecolor="black",
    range=(precio_min_poblacion_mas_frecuente_no_cero, precio_max_poblacion_mas_frecuente)
)
plt.title("Histograma de Precios en" + poblacion_mas_frecuente)
plt.xlabel("Precio (USD)")
plt.ylabel("Frecuencia")


plt.tight_layout()
plt.show()

# Trazar el histograma de precios
```
